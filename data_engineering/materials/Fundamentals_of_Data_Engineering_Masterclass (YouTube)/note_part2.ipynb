## Data Architecture

Data Architecture is the design of system to support evolving data needs of an enterprise, achieved by flexible and reversible decisions reached through a careful evaluation of trade-offs.

### 1. Business Needs (Operational Architecture)

### Focus on the business goals and requirement - Why  
  * Start with the End in Mind
  * Iterate and Evolve
  * Focus on Impact

### 2. Technology Integration (Technical Architecture)

Focus on the technical side of things, and how to ingest, store, and transform data - How  
  * Simplicity is Key
  * Choose the Right Tools for the Job
  * Build for Scale and Flexibility
  * Embrace Automation
  * Prioritize Data Security and Governance

## Slow Changing Dimension  
Dimension data may change over time. Do we need to keep history record?

### Designing SCD1
Values are overwriten and not history is maintained.

### Designing SCD2
We maintain a complete history of changes. Every time there is a change, we add a new row with all the details without deleting the previous values.
  * Using a flag (e.g. isActive column)
  * Using Version numbers
  * Using date ranges (e.g. startDate/endDate columns)

### Designing SCD3
We maintain only a partial history and not a complete history. Instead of adding additional rows, we add an extra column that stores the previous value, so only one version of historic data wil be preserved.

### Designing SCD6
A combination of 1, 2 and 3. Along with the additino of new rows, we also update the latest value in all the rows.

## Data Marts
A subset of a data warehouse.  
A company may have several departments and many tables in its data warehouse. Each department may not need all the information in data warehouse, so it will collect (some) columns from (some) tables to build its own table, or reporting system.

## Data Lake


### Schema-on-read
Schema-on-read means data is stored in its raw format first, and the schema is applied only when the data is read or queried.
It is commonly used in data lakes because it provides flexibility for handling structured, semi-structured, and unstructured data.

## Data Lake vs Data Warehouse

|        | Data Warehouse  | Data Lake      |
| ------ | --------------- | -------------- |
| Schema | Schema-on-Write | Schema-on-Read |
| 数据类型   | 结构化             | 任意类型           |
| 灵活性    | 较低              | 高              |
| 数据质量   | 高               | 容易混乱           |
| 查询性能   | 快               | 较慢             |
