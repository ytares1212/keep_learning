# Data Architecture

Data Architecture is the design of system to support evolving data needs of an enterprise, achieved by flexible and reversible decisions reached through a careful evaluation of trade-offs.

## 1. Business Needs (Operational Architecture)

### Focus on the business goals and requirement - Why  
  * Start with the End in Mind
  * Iterate and Evolve
  * Focus on Impact

## 2. Technology Integration (Technical Architecture)

Focus on the technical side of things, and how to ingest, store, and transform data - How  
  * Simplicity is Key
  * Choose the Right Tools for the Job
  * Build for Scale and Flexibility
  * Embrace Automation
  * Prioritize Data Security and Governance
