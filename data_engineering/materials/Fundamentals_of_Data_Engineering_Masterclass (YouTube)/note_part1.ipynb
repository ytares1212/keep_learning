# The Data Engineering Lifecycle 


* **Generation** -> Raw Data
* **Ingestion** -> Making a connection in such a way that any time a new data is getting generated, we automatically fetch this data and store it
* **Transformation** -> Data coming from all sources passes through a certain transformation logic so that our data gets structured
* **Serving**: Serve data to user i.e. ML, DA etc,.

# Data Modelling
**SQL Database**
1. Consider what kind of data does a company/business need to collect? For example: Amazon - Orders, Users, Payments, shipppings, Products, Sellers etc
2. Each type of data will have its own table
3. These tables need to be joined together by some common keys, they must related to some of others in some way
4. Each single column has some kind of schema or data type attach to it, this is called SQL or relational database table

**NoSQL Database**  
Data can be stored in different type of format:
* Key-Value
* Column-Family
* Graph
* Document

# Data Storage Processing
**OLTP**: Online Transactional Processing (**Relational DB**)
* Process Transactional Data
* Good for row level operation: Fast insert, update, read
* mostly row based

**OLAP**: Online Analytical Processing (**Data Warehouse**)
* Analysis Workload
* mostly column based

**OLTP -> Transformation -> OLAP**

# ETL Processing

## Extract
* from difference sources e.g. RDBMS, APIs etc
## Transformation
* apply some logic e.g removing duplicates, handling null values etc
## Load
* to Data Warehouse e.g Snowflake DB, BigQuery etc
* to Object Storage e.g S3, Google Cloud Storage, Azure Data Lake etc
