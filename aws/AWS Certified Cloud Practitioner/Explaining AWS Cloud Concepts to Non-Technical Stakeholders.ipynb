### 1. Describe what cloud computing is and the benefits of using AWS for businesses.  
* Cloud computing is the on-demand delivery of compute power, databse storage, applications and other IT resources.
* Benefits:
  * On-demand self service:
    * Users can provision resources and use them without human interaction from the service
provider
  * Broad network access:
    * Resources available over the network, and can be accessed by diverse client platforms
  * Multi-tenancy and resource pooling:
    * Multiple customers can share the same infrastructure and applications with security and privacy
    * Multiple customers are serviced from the same physical resources
  * Rapid elasticity and scalability:
    * Automatically and quickly acquire and dispose resources when needed
    * Quickly and easily scale based on demand
  * Measured service:
    * Usage is measured, users pay correctly for what they have used

Cloud computing is essentially like using electricity instead of owning a generator. Instead of buying and maintaining your own IT infrastructure, you access computing resources over the internet, only when you need them.  
For a business, this means:
* You don’t need to invest heavily upfront in servers and data centres
* You can scale resources up or down depending on demand
* Your teams can access systems from anywhere
* And you only pay for what you actually use  

AWS provides this capability at a global scale, allowing even smaller companies to access enterprise-level infrastructure without the complexity of managing it themselves.  
In short, cloud computing helps businesses become more flexible, cost-efficient, and faster to respond to change

### 2. Effectively compare traditional on-premises infrastructure with AWS in a straightforward manner.  
* Traditional on-premises infrastructure:
  * Pay the rent, power supply, cooling, maintenance for the data center
  * Adding and replacing hardware takes time
  * Scaling is limited
  * Need to hire 24/7 team to monitor the infrastructure
  * Hard to deal with disasters (earthquake, power shutdown, fire etc)

* AWS:
  * Trade capital expense (CAPEX) for operational expense (OPEX)
    * Pay On-Demand: don’t own hardware
    * Reduced Total Cost of Ownership (TCO) & Operational Expense (OPEX)
  * Benefit from massive economies of scale
    * Prices are reduced as AWS is more efficient due to large scale
  * Stop guessing capacity
    * Scale based on actual measured usage
  * Increase speed and agility
  * Stop spending money running and maintaining data centers
  * Go global in minutes: leverage the AWS global infrastructure


Traditionally, companies would need to purchase and maintain their own servers, which involves upfront capital investment, ongoing maintenance, and planning for peak capacity—even if that capacity is not always used.  
With AWS, this model changes significantly:
* Instead of large upfront costs, you pay as you go
* Instead of planning for maximum capacity, you scale based on actual demand
* Instead of managing hardware, your team can focus on core business activities  

This reduces capital and operational burden and allows the business to be more agile, especially when dealing with growth or unexpected changes in demand.

### 3. Explain the shared responsibility model, ensuring the security responsibilities of both AWS and the customer are clear.
* CUSTOMER = RESPONSIBILITY FOR THE SECURITY **IN** THE CLOUD
* AWS = RESPONSIBILITY FOR THE SECURITY **OF** THE CLOUD



When it comes to security, AWS works on a shared responsibility model.  
AWS is responsible for securing the underlying infrastructure — such as data centres, physical servers, and the core cloud platform.  
Your company remains responsible for how you use those services — for example, who has access to your systems, how your data is managed, and your internal security policies.  
A simple way to think about it is: AWS provides a highly secure building, and you decide who gets the keys to your office inside it.  
This model ensures that security is a shared effort, combining AWS’s large-scale investment in infrastructure security with your control over business data and access.

From a business perspective, AWS helps reduce large upfront infrastructure costs by moving to a pay-as-you-go model, so you only pay for what you use. This avoids over-investment and improves cost control, especially when demand fluctuates.

In terms of security, AWS operates highly secure global data centres and invests far more in security than most companies can individually. At the same time, you still maintain control over your data and access.

So overall, AWS not only improves cost efficiency but also strengthens operational reliability and security.
